### Build a tally of a list four ways:
1. if event in counts: ... else: ...
2. counts[event] = counts.get(event, 0) + 1
3. defaultdict(int) with counts[event] += 1
4. Counter(log) <br><br>
Complexity: time O(n), space O(k), where k is the number of distinct events.

```python
log = ["grasp", "lift", "grasp", "drop", "grasp"]
```

```python
# 1. pure if then else
counts = {}
for action in log:
    if action in counts:
        counts[action] = counts[action] + 1
    else:
        counts[action] = 1

print(counts)
```

```python
# 2. Using .get()

counts = {}
for action in log:
    counts[action] = counts.get(action, 0) + 1

print(counts)
```

```python
# 3. Using defaultdict()

from collections import defaultdict

counts = defaultdict(int)
for action in log:
    counts[action] = counts[action] + 1

print(counts)
```

```python
#4. using Counter()

from collections import Counter

counts = Counter(log)
print(counts)
```

### most_common

Counter also has a method that answers “which events happened most?”:

```python
print(counts.most_common(1))
print(counts.most_common(2))
```

```python
top_1 = counts.most_common(1)[0][0]
print(top_1)
```

### Word frequency on a string

The robot’s log arrives as one string, not a list. Two string tools get you back to the list you already know:

```python
from collections import Counter

text = "grasp, lift grasp" #"Grasp lift GRASP drop grasp"
words = text.lower().replace(',',' ').split()
counts = Counter(words)
assert Counter(words) == {"grasp": 2, "lift": 1}
```

### Group anagrams. 
Anagrams are words made of the same letters, like "arm" and "mar". The goal: group them, e.g. ["arm", "mar", "lift", "rma"] becomes [["arm", "mar", "rma"], ["lift"]].

```python
from collections import defaultdict

by_joint = defaultdict(list)
by_joint["elbow"].append(1.2)
by_joint["elbow"].append(1.5)
by_joint["wrist"].append(-0.3)
print(by_joint)   # defaultdict(list, {'elbow': [1.2, 1.5], 'wrist': [-0.3]})
```

```python
readings = [("elbow", 1.2), ("wrist", -0.3), ("elbow", 1.5)]

joints = defaultdict(list)
for joint, reading in readings:
    joints[joint].append(reading)

assert joints == {"elbow": [1.2, 1.5], "wrist": [-0.3]}
```
