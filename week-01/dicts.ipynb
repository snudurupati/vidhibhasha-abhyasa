### Build a tally of a list four ways:
1. if event in counts: ... else: ...
2. counts[event] = counts.get(event, 0) + 1
3. defaultdict(int) with counts[event] += 1
4. Counter(log) <br><br>
Complexity: time O(n), space O(k), where k is the number of distinct events.

```python
log = ["grasp", "lift", "grasp", "drop", "grasp"]
```

```python
# 1. pure if then else
counts = {}
for action in log:
    if action in counts:
        counts[action] = counts[action] + 1
    else:
        counts[action] = 1

print(counts)
```

```python
# 2. Using .get()

counts = {}
for action in log:
    counts[action] = counts.get(action, 0) + 1

print(counts)
```

```python
# 3. Using defaultdict()

from collections import defaultdict

counts = defaultdict(int)
for action in log:
    counts[action] = counts[action] + 1

print(counts)
```

```python
#4. using Counter()

from collections import Counter

counts = Counter(log)
print(counts)
```

### most_common

Counter also has a method that answers “which events happened most?”:

```python
print(counts.most_common(1))
print(counts.most_common(2))
```

```python
top_1 = counts.most_common(1)[0][0]
print(top_1)
```

### Word frequency on a string

The robot’s log arrives as one string, not a list. Two string tools get you back to the list you already know:

```python
from collections import Counter

text = "grasp, lift grasp" #"Grasp lift GRASP drop grasp"
words = text.lower().replace(',',' ').split()
counts = Counter(words)
assert Counter(words) == {"grasp": 2, "lift": 1}
```

### Pattern: canonical key grouping. 
Group items by a standard form that ignores differences that don’t matter.

The recipe<br>

1. Decide which differences to ignore (case, order, timestamps).
2. Write a function that turns any item into one standard form, the canonical key.
3. Group by that key with defaultdict(list) and .append.

Worked example: joint names with inconsistent case.

```python
names = ["Elbow", "WRIST", "elbow", "wrist", "Elbow"]
groups = defaultdict(list)
for name in names:
    groups[name.lower()].append(name)
# {"elbow": ["Elbow", "elbow", "Elbow"], "wrist": ["WRIST", "wrist"]}
print(groups)
```

Task: Group `["arm", "mar", "lift", "rma"]` into `[["arm", "mar", "rma"], ["lift"]]`.

```python
words = ["arm", "mar", "lift", "rma"]
groups = defaultdict(list)

for word in words:
    key = "".join(sorted(word.lower()))
    groups[key].append(word)

result = list(groups.values())
assert result == [["arm", "mar", "rma"], ["lift"]]
```
